```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("berkanoztas/synthetic-transaction-monitoring-dataset-aml")

print("Path to dataset files:", path)
```

```python
path+'/SAML-D.csv'
```

```python
import pandas as pd
data=pd.read_csv(path+'/SAML-D.csv')
```

```python
data[:2]
```

```python
df=data.drop(['Time','Date','Laundering_type'],axis=1)
```

```python
df[:2]
```

```python
from collections import Counter
Counter(df['Is_laundering'])
```

```python
from sklearn.preprocessing import LabelEncoder
df1=df.apply(LabelEncoder().fit_transform)
```

```python
df1[:2]
```

```python
target=df1['Is_laundering']
features=df1.drop(['Is_laundering'],axis=1)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(features,target,test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
pred=lr.predict(X_test)
```

```python
Counter(pred)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,pred))
```
