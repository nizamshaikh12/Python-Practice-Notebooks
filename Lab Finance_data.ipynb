```python
import pandas as pd

# Load the data
data = pd.read_csv('/content/Finance_data.csv')

data[:3]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'column_name' is a numerical column in your data
sns.histplot(data['age'])
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'column1' and 'column2' are numerical columns
sns.scatterplot(x='age', y='Mutual_Funds', data=data)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Assuming 'categorical_column' and 'numerical_column' are relevant columns
sns.boxplot(x='Avenue', y='age', data=data)
plt.show()
```

```python
#Show frequency age range by gender
data.groupby('gender').age.plot(kind='kde')
plt.xlabel('age')
plt.legend()
plt.show()
```

```python
from wordcloud import WordCloud

# Assuming 'text_column' is the column containing text data
text = ' '.join(data['Source'].astype(str).tolist())

# Create the word cloud

wordcloud = WordCloud(width = 800, height = 800,
                background_color ='white',
                min_font_size = 10).generate(text)

# Display the word cloud
plt.figure(figsize = (4, 4), facecolor = None)
plt.imshow(wordcloud)
plt.axis("off")
plt.tight_layout(pad = 0)
plt.show()
```

```python

from collections import Counter
# Assuming 'text_column' is the column containing text data
text = ' '.join(data['Reason_FD'].astype(str).tolist())

# Split the text into words and count their occurrences
words = text.lower().split()
word_counts = Counter(words)

# Get the most common words and their counts
top_words = word_counts.most_common(10)  # Get top 10 words (adjust as needed)
word_labels, word_frequencies = zip(*top_words)

# Create the bar chart
plt.figure(figsize=(10, 6))  # Adjust figure size as needed
plt.bar(word_labels, word_frequencies)
plt.title('Top Word Frequencies')
plt.xlabel('Words')
plt.ylabel('Frequency')
plt.xticks(rotation=45, ha='right')  # Rotate x-axis labels for better readability
plt.tight_layout()
plt.show()
```
