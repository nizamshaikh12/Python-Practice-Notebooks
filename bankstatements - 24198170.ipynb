**Read Bank Statment File**

```python
import pandas as pd
df = pd.read_csv("bankstatements.csv")
```

```python
df.head()
```

**Task 1: Selecting Specific Columns**

Select the 'date' and 'amount' columns

```python
# Select the 'date' and 'amount' columns
SC = df[["date", "amount"]]
SC
```

**Task 2: Selecting Rows Based on a Condition**

Select transactions where the amount is greater than 1000


```python
# Select transactions where the amount is greater than 1000
SR = df[df["amount"] > 1000]
SR
```

**Task 3: Selecting Rows Based on Multiple Conditions**

Select transactions in January 2022 with amount > 5000

```python
# Select transactions in January 2022 with amount > 5000
SRM = df[(df["Month"] == 1) & (df["Year"] == 2022) & (df["amount"] > 5000)]
SRM
```

**Task 4:Sort transactions by date in ascending order**

```python
# Sort transactions by date in ascending order
SD = df.sort_values(by=["date"], ascending=True)
SD
```

**Task 5: Calculate the total amount of all transactions**

```python
# Calculate the total amount of all transactions
TA = df["amount"].sum()
TA
```

**Task 6: Grouping Data**

```python
# Group transactions by mode and calculate the average amount for each mode
GD = df.groupby("mode")["amount"].mean()
GD
```

**Task 7: Creating New Columns**

```python
# Create a new column 'transaction_type' based on 'DrCr'
df["transaction_type"] = df["DrCr"].map({"Db": "Debit", "Cr": "Credit"})
df.head()
```

**Task 8: Renaming Columns**

```python
# Rename the 'amount' column to 'transaction_amount'
df = df.rename(columns={"Tday": "Transaction_Day"})
df.head()
```

**Task 9: Dropping Columns**

```python
# Drop the 'Tday' column
df = df.drop(columns=["transaction_type"])
df
```

**Task 10: Handling Missing Values**

```python
# Fill missing values in the 'name' column with 'Unknown'
df["name"] = df["name"].fillna("Nizam")
df
```

**Task 11: Filtering Based on String Contains**

```python
# Select transactions where the name contains 'AYUB'
FS = df[df["name"].str.contains("Nizam")]
FS
```

**Task 12: Extracting Date Components**

```python
# Extract the day of the week from the 'date' column
df["date"] = pd.to_datetime(df["date"], dayfirst=True)
df["week_name_days"] = df["date"].dt.day_name()
df
```

**Task 13: Applying a Function to a Column**

```python
# Apply a function to double the 'amount' column
df["Multiply_amount"] = df["amount"].apply(lambda x: x * 2)
df
```

**Task 14: Counting Unique Values**

```python
# Count the number of unique transaction modes
unique_values = df['mode'].nunique()
unique_values
```

**Task 15: Checking for Duplicates**

```python
# Find and display duplicate rows
duplicates = df[df.duplicated()]
duplicates
```
